```python
#import pandas as pd
import wbgapi as wb
import country_converter as coco

# --- 1. Fetch World Bank Data (2022) ---
# 'SP.POP.TOTL' is Total Population
# time=2022 fetches only that year
# mrnev=1 is a fallback to the 'most recent non-empty value' if 2022 is missing
print("Fetching data from World Bank...")
econ_gen = wb.data.fetch(['SP.POP.TOTL','NY.GDP.PCAP.CD','SI.POV.GINI', 'HD_HCIP_OVRL_TO', 'AG.LND.TOTL.K2', 'EN.POP.DNST'], time=2022, mrnev=1, labels=True )
econ_gen
```

```python
# 2. In ein temporäres DataFrame umwandeln
df_temp = pd.DataFrame(econ_gen)
df_temp
```

```python
# 2. Gezieltes Auslesen der Dictionary-Werte
# Series: 'value' -> wird Spaltenname (z.B. "GDP (current US$)")
df_temp['series'] = df_temp['series'].apply(lambda x: x['value'] if isinstance(x, dict) else x)

# Economy: 'id' -> wird ISO3 (z.B. "AUT")
df_temp['economy'] = df_temp['economy'].apply(lambda x: x['id'] if isinstance(x, dict) else x)

# Time: 'value' -> wird das Jahr (z.B. "2022")
df_temp['time'] = df_temp['time'].apply(lambda x: x['value'] if isinstance(x, dict) else x)
```

```python
import re

# 1. Die Reinigungs-Funktion (nur für den Basis-Namen)
def clean_base_name(name):
    if not isinstance(name, str):
        return name
    # Alles vor '(', ':', '[' oder ',' abschneiden (Regex erweitert)
    base_name = re.split(r'[()|:|\[|,]', name)[0]
    # Trimmen und Blanks durch Underscores ersetzen
    return base_name.strip().replace(' ', '_')

# 2. Anwendung auf df_temp (bevor pivot)
df_temp['series'] = df_temp['series'].apply(clean_base_name)

# 3. Pivotieren
df_values = df_temp.pivot(index='economy', columns='series', values='value')
df_years = df_temp.pivot(index='economy', columns='series', values='time')

# 4. Suffixe mit eckigen Klammern hinzufügen
# Hier sorgen wir dafür, dass deine gewünschten Labels [Value] und [Year] dranbleiben
df_values = df_values.add_suffix('_[Value]')
df_years = df_years.add_suffix('_[Year]')

# 5. Zusammenführen
df_wb_final = pd.concat([df_values, df_years], axis=1).reset_index()
df_wb_final = df_wb_final.rename(columns={'economy': 'ISO3'})

df_wb_final
```

```python
# 1. Alle Spalten finden, die auf '[Year]' enden
year_cols = [col for col in df_wb_final.columns if col.endswith('_[Year]')]

# 2. Minimum und Maximum berechnen
# Wir wandeln die Werte kurz in Zahlen um (pd.to_numeric), damit 2010 < 2022 korrekt erkannt wird
year_stats = df_wb_final[year_cols].apply(pd.to_numeric, errors='coerce').agg(['min', 'max'])

print("Zeitspanne der Weltbank-Daten pro Indikator:")
print(year_stats)
```

```python
# 1. Alle Spalten entfernen, die '[Year]' enthalten
# Wir behalten nur die Spalten, die NICHT auf '_[Year]' enden
cols_to_keep = [col for col in df_wb_final.columns if not col.endswith('_[Year]')]
df_wb_final = df_wb_final[cols_to_keep]

# 2. Den Teil '_[Value]' aus den verbleibenden Spaltennamen löschen
# .str.replace funktioniert hier direkt auf der Liste der Spaltennamen
df_wb_final.columns = df_wb_final.columns.str.replace('_[Value]', '', regex=False)

# Kurze Kontrolle der neuen Spaltenstruktur
print(df_wb_final.columns.tolist())
```

To mitigate the 'Law of Small Numbers' and the resulting high statistical variance, countries with a population of fewer than 200,000 inhabitants were excluded from this analysis. In very small populations, indivisible infrastructure units—such as individual schools—create mathematical artifacts in density calculations. These outliers often reflect the inherent volatility of small sample sizes rather than actual structural or systemic superiority.

This approach follows the principle that the standard error of a sample mean is inversely proportional to the square root of the population size; hence, smaller units exhibit a disproportionately high noise-to-signal ratio.

```python
# Filter setzen: Nur Länder mit mehr als 200k Einwohnern behalten
min_pop = 200000
df_filtered = final_df[final_df['Population'] >= min_pop].copy()

# Liste der entfernten Länder für deinen Report
dropped_countries = final_df[final_df['Population'] < min_pop]['Country'].tolist()
print(f"Ausgeschlossene Kleinststaaten ({len(dropped_countries)}): {dropped_countries}")

final_df = df_filtered.copy()
```

```python
# 1. Sicherstellen, dass die Spalten numerisch sind
final_df['Count'] = pd.to_numeric(final_df['Count'], errors='coerce')
final_df['Population'] = pd.to_numeric(final_df['Population'], errors='coerce')

# 2. Berechnung: Schulen pro 100.000 Einwohner
# Formel: (Anzahl Schulen / Bevölkerung) * 1.000.000
final_df['School_density'] = (final_df['Count'] / final_df['Population']) * 1000000

# 3. Kleiner Check: Die Top 5 Länder nach Schuldichte
print(final_df[['Country', 'Count', 'Population', 'School_density']].sort_values(by='School_density', ascending=False).head(20))
```

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import numpy as np
from scipy import stats

df_analysis = final_df[final_df['Count']>10].copy()

# 1. Parameter definieren
x_params = ['GDP_per_capita', 'Gini_index', 'Human_capital_index_plus', 'Population_density']
titles = [f"School Density vs. {p.replace('_', ' ')}" for p in x_params]

# 2. Subplot-Layout (2x2)
fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=titles,
    vertical_spacing=0.15
)

for i, x_col in enumerate(x_params):
    row = (i // 2) + 1
    col = (i % 2) + 1
    
    # Daten säubern (Entferne NaNs für die Regression)
    mask = df_analysis[x_col].notnull() & df_analysis['School_density'].notnull()
    df_plot = df_analysis[mask]
    
    x = df_plot[x_col]
    y = df_plot['School_density']
    
    # --- Lineare Regression berechnen ---
    slope, intercept, r_value, p_value, std_err = stats.linregress(x, y)
    line = slope * x + intercept
    
    # 3. Scatter Plot (Die Länder-Punkte)
    fig.add_trace(
        go.Scatter(
            x=x, y=y,
            mode='markers',
            name='Datenpunkte',
            text=df_plot['Country'],
            marker=dict(color='rgba(100, 150, 250, 0.6)', size=8),
            showlegend=False
        ),
        row=row, col=col
    )
    
    # 4. Regressionsgerade hinzufügen
    fig.add_trace(
        go.Scatter(
            x=x, y=line,
            mode='lines',
            name='Regression',
            line=dict(color='red', width=2),
            showlegend=False
        ),
        row=row, col=col
    )
    
    # 5. Korrelationsindex (R) als Annotation auf die Linie schreiben
    # Wir platzieren den Text in der Mitte der Linie
    mid_x = x.mean()
    mid_y = slope * mid_x + intercept
    
    fig.add_annotation(
        x=mid_x, y=mid_y,
        text=f"R = {r_value:.2f}",
        showarrow=True,
        arrowhead=1,
        ax=40, ay=-40,
        bgcolor="white",
        bordercolor="red",
        row=row, col=col
    )

    # Achsen beschriften
    fig.update_xaxes(title_text=x_col.replace('_', ' '), row=row, col=col)
    fig.update_yaxes(title_text="Schul-Dichte (pro Million)", row=row, col=col)

# Layout-Anpassungen
fig.update_layout(
    height=900, width=1100,
    title_text="Korrelations-Analyse: Was treibt die Schul-Dichte?",
    template="plotly_white"
)

fig.show()
```

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import numpy as np
from scipy import stats

# 1. Parameter definieren
x_params = ['GDP_per_capita', 'Gini_index', 'Human_capital_index_plus', 'Land_area']
log_x_params = ['GDP_per_capita', 'Land_area'] 

# 2. Subplot-Layout (2x2)
fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=[f"vs. {p.replace('_', ' ')}" for p in x_params],
    vertical_spacing=0.15,
    horizontal_spacing=0.12
)

for i, x_col in enumerate(x_params):
    row = (i // 2) + 1
    col = (i % 2) + 1
    
    # Daten säubern
    lower_limit = df_analysis['School_density'].quantile(0.1)
    mask = df_analysis[x_col].notnull() & df_analysis['School_density'].notnull() & (df_analysis[x_col] > 0)# & (df_analysis['School_density'] > lower_limit)
    df_plot = df_analysis[mask].copy()
    
    x_data = df_plot[x_col]
    y_data = df_plot['School_density']
    
    # Regression berechnen (X log-transformieren falls nötig)
    x_reg = np.log10(x_data) if x_col in log_x_params else x_data
    slope, intercept, r_value, p_value, std_err = stats.linregress(x_reg, y_data)
    
    # Regressionslinie
    x_range = np.linspace(x_reg.min(), x_reg.max(), 100)
    y_line = slope * x_range + intercept
    x_line_plot = 10**x_range if x_col in log_x_params else x_range

    # 3. Scatter Plot
    fig.add_trace(
        go.Scatter(x=x_data, y=y_data, mode='markers', text=df_plot['Country'],
                   marker=dict(color='rgba(100, 150, 250, 0.6)', size=8), showlegend=False),
        row=row, col=col
    )
    
    # 4. Regressionslinie
    fig.add_trace(
        go.Scatter(x=x_line_plot, y=y_line, mode='lines', line=dict(color='red', width=2), showlegend=False),
        row=row, col=col
    )
    
    # 5. KORREKTUR: Annotation mit dynamischer Achsen-Referenz
    # Plotly Regel: Die erste Achse heißt oft nur 'x domain', nicht 'x1 domain'
    axis_idx = i + 1
    xref_str = f"x{axis_idx} domain" if axis_idx > 1 else "x domain"
    yref_str = f"y{axis_idx} domain" if axis_idx > 1 else "y domain"

    fig.add_annotation(
        xref=xref_str, yref=yref_str,
        x=0.05, y=0.95,
        text=f"<b>R = {r_value:.2f}</b><br>{'Log' if x_col in log_x_params else 'Lin'}-Lin",
        showarrow=False, bgcolor="white", bordercolor="black", align="left",
        row=row, col=col
    )

    # 6. Achsen-Skalierung & Beschriftung
    if x_col in log_x_params:
        fig.update_xaxes(type="log", row=row, col=col)
    
    fig.update_xaxes(title_text=x_col.replace('_', ' '), row=row, col=col)
    fig.update_yaxes(title_text="School Density (linear)", row=row, col=col)

# Finales Layout
fig.update_layout(height=900, width=1200, title_text="School Density Analysis: Final Dashboard", template="plotly_white")
fig.show()
```

"To further refine the dataset, we excluded the bottom 10th percentile of countries by school density. These extreme low-performers likely represent 'structural outliers' where the density is not driven by macroeconomic factors but by unobserved institutional barriers. Such barriers may include restrictive legal frameworks for non-governmental school establishment or a lack of integration into international educational networks (e.g., the Montessori community). By removing this bottom decile, we minimize the impact of non-systemic 'zeros' and ensure that our correlation analysis focuses on countries with functional and comparable educational infrastructure."

```python
import statsmodels.api as sm

# 1. Daten vorbereiten (Log-Transformationen wie besprochen)
df_model = df_analysis.dropna(subset=['GDP_per_capita', 'Gini_index', 'Human_capital_index_plus', 'Land_area', 'School_density'])

X = df_model[['GDP_per_capita', 'Gini_index', 'Human_capital_index_plus', 'Land_area']].copy()
X['GDP_per_capita'] = np.log10(X['GDP_per_capita'])
X['Land_area'] = np.log10(X['Land_area'])
X = sm.add_constant(X) # Intercept hinzufügen

y = df_model['School_density']

# 2. Modell rechnen
model = sm.OLS(y, X).fit()

# 3. Residuen speichern
df_model['Predicted'] = model.predict(X)
df_model['Residual'] = df_model['School_density'] - df_model['Predicted']

# 4. Die "System-Blockierer" finden
# Länder, deren Dichte viel NIEDRIGER ist als erwartet (negativstes Residuum)
blockierer = df_model.sort_values(by='Residual').head(20)

print("Länder, die trotz guter Parameter 'unten bleiben':")
print(blockierer[['Country', 'School_density', 'Predicted', 'Residual']])
```

```python
import pandas as pd
import numpy as np
from scipy import stats
import statsmodels.api as sm

# --- Schritt A: Das Residuen-Modell zur Identifikation der Blockierer ---
# Wir nutzen alle 4 Parameter, um zu sehen, wer sich nicht an die Regeln hält
df_res = df_model.dropna(subset=['GDP_per_capita', 'Gini_index', 'Human_capital_index_plus', 'Land_area', 'School_density']).copy()

X = df_res[['GDP_per_capita', 'Gini_index', 'Human_capital_index_plus', 'Land_area']].copy()
X['GDP_per_capita'] = np.log10(X['GDP_per_capita'])
X['Land_area'] = np.log10(X['Land_area'])
X = sm.add_constant(X)
y = df_res['School_density']

model = sm.OLS(y, X).fit()
df_res['Residual'] = model.resid

# Die 20 Länder mit den negativsten Residuen (die "Blockierer")
# 1. Die potenziellen Blockierer identifizieren (wie bisher)
blockierer_kandidaten = df_res.sort_values(by='Residual').head(20)

# 2. Den Filter anwenden: 
# Wir nehmen nur die Länder in die "Abschussliste", die AUCH eine niedrige Dichte haben.
# Länder mit School_density > 3 werden "begnadigt".
tatsaechliche_blockierer = blockierer_kandidaten[blockierer_kandidaten['School_density'] <= 3]
blockierer_namen = tatsaechliche_blockierer['Country'].tolist()

# 3. Datensatz bereinigen
df_meaningful = df_res[~df_res['Country'].isin(blockierer_namen)].copy()

print(f"Identifizierte Blockierer-Kandidaten: 20")
print(f"Davon 'begnadigt' (Dichte > 3): {len(blockierer_kandidaten) - len(tatsaechliche_blockierer)}")
print(f"Endgültig entfernte Länder: {len(blockierer_namen)}")
print(f"Verbleibende Länder: {len(df_meaningful)}")

# --- Schritt B: Datensatz bereinigen ---
# Wir erstellen einen neuen DataFrame ohne diese 20 Länder
df_meaningful = df_res[(~df_res['Country'].isin(blockierer_namen))].copy()

print(f"Entfernte Blockierer: {blockierer_namen}")
print(f"Verbleibende Länder für finale Korrelation: {len(df_meaningful)}")

# --- Schritt C: Finale R-Werte berechnen ---
final_r = {}
for x_col in ['GDP_per_capita', 'Gini_index', 'Human_capital_index_plus', 'Land_area']:
    # Gleiche Logik wie im Dashboard (Log-Skala für GDP und Fläche)
    x_reg = np.log10(df_meaningful[x_col]) if x_col in ['GDP_per_capita', 'Land_area'] else df_meaningful[x_col]
    r_val, _ = stats.pearsonr(x_reg, df_meaningful['School_density'])
    final_r[x_col] = r_val

# Ergebnisse ausgeben
print("\n--- Finale Korrelationskoeffizienten (ohne Blockierer) ---")
for param, r in final_r.items():
    print(f"{param:25}: R = {r:.3f}")
```

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import numpy as np
from scipy import stats

# 1. Parameter für die 4 Grafiken
x_params = ['GDP_per_capita', 'Gini_index', 'Human_capital_index_plus', 'Land_area']
log_x_params = ['GDP_per_capita', 'Land_area']  # Diese erhalten eine Log-X-Achse

# 2. Subplot-Layout erstellen
fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=[f"vs. {p.replace('_', ' ')}" for p in x_params],
    vertical_spacing=0.15,
    horizontal_spacing=0.12
)

for i, x_col in enumerate(x_params):
    row = (i // 2) + 1
    col = (i % 2) + 1
    
    # Daten für diesen spezifischen Plot (NaNs in X ausschließen)
    df_plot = df_meaningful[df_meaningful[x_col].notnull() & (df_meaningful[x_col] > 0)].copy()
    
    x_data = df_plot[x_col]
    y_data = df_plot['School_density']
    
    # --- Regression berechnen ---
    # Log-Transformation für X falls in log_x_params, Y bleibt linear
    x_reg = np.log10(x_data) if x_col in log_x_params else x_data
    slope, intercept, r_value, p_value, std_err = stats.linregress(x_reg, y_data)
    
    # Regressionslinie für den Plot (100 Punkte für eine glatte Kurve im Log-Raum)
    x_range_reg = np.linspace(x_reg.min(), x_reg.max(), 100)
    y_line = slope * x_range_reg + intercept
    x_line_plot = 10**x_range_reg if x_col in log_x_params else x_range_reg

    # 3. Scatter Plot hinzufügen
    fig.add_trace(
        go.Scatter(
            x=x_data, y=y_data,
            mode='markers',
            text=df_plot['Country'],
            hovertemplate="<b>%{text}</b><br>" + x_col + ": %{x}<br>Dichte: %{y}<extra></extra>",
            marker=dict(
                color='rgba(63, 81, 181, 0.7)', 
                size=9,
                line=dict(width=1, color='white')
            ),
            showlegend=False
        ),
        row=row, col=col
    )
    
    # 4. Regressionslinie hinzufügen
    fig.add_trace(
        go.Scatter(
            x=x_line_plot, y=y_line,
            mode='lines',
            line=dict(color='firebrick', width=3),
            showlegend=False
        ),
        row=row, col=col
    )
    
    # 5. Annotation (R-Wert) - mit dynamischer Achsenreferenz für Plotly-Kompatibilität
    axis_idx = i + 1
    xref_str = f"x{axis_idx} domain" if axis_idx > 1 else "x domain"
    yref_str = f"y{axis_idx} domain" if axis_idx > 1 else "y domain"

    fig.add_annotation(
        xref=xref_str, yref=yref_str,
        x=0.05, y=0.95,
        text=f"<b>R = {r_value:.3f}</b><br>p = {p_value:.4f}",
        showarrow=False,
        bgcolor="rgba(255,255,255,0.9)",
        bordercolor="gray",
        align="left",
        row=row, col=col
    )

    # 6. Achsen-Konfiguration
    if x_col in log_x_params:
        fig.update_xaxes(type="log", row=row, col=col)
    
    fig.update_xaxes(title_text=x_col.replace('_', ' '), row=row, col=col)
    fig.update_yaxes(title_text="School Density (linear)", row=row, col=col)

# Finales Layout-Styling
fig.update_layout(
    height=1000, width=1200,
    title_text="<b>Final Analysis: Structural Drivers of School Density</b><br><sup>Excluding structural 'blocker' countries (Residual-based filtering)</sup>",
    template="plotly_white",
    font=dict(family="Arial", size=12)
)

fig.show()
```

```python
import plotly.graph_objects as go
import numpy as np
from scipy import stats
from sklearn.preprocessing import MinMaxScaler

# --- 1. Frische Korrelationen (Gewichte) berechnen ---
df_idx = df_meaningful.copy()
factors = ['GDP_per_capita', 'Gini_index', 'Human_capital_index_plus', 'Land_area']
weights = {}

for col in factors:
    # Transformation wie im Modell (Log für GDP und Area)
    x_val = np.log10(df_idx[col]) if col in ['GDP_per_capita', 'Land_area'] else df_idx[col]
    r, _ = stats.pearsonr(x_val, df_idx['School_density'])
    weights[col] = abs(r) # Wir nutzen den Betrag der Korrelation als Gewicht

print(f"Genutzte Gewichte (R-Beträge): {weights}")

# --- 2. Daten für den Index aufbereiten ---
# Invertierung: Niedriger Gini und kleine Fläche sollen den Index ERHÖHEN
df_idx['Gini_inv'] = df_idx['Gini_index'].max() - df_idx['Gini_index']
df_idx['Area_inv'] = np.log10(df_idx['Land_area']).max() - np.log10(df_idx['Land_area'])
df_idx['GDP_log'] = np.log10(df_idx['GDP_per_capita'])

# Normalisierung (0 bis 1)
scaler = MinMaxScaler()
cols_to_scale = ['GDP_log', 'Gini_inv', 'Human_capital_index_plus', 'Area_inv']
df_idx[cols_to_scale] = scaler.fit_transform(df_idx[cols_to_scale])

# --- 3. Den Kombinierten Potenzial-Index berechnen ---
total_w = sum(weights.values())
df_idx['Combined_Potential_Index'] = (
    (df_idx['GDP_log'] * weights['GDP_per_capita'] +
     df_idx['Gini_inv'] * weights['Gini_index'] +
     df_idx['Human_capital_index_plus'] * weights['Human_capital_index_plus'] +
     df_idx['Area_inv'] * weights['Land_area']) / total_w
)

# --- 4. Bubble Size Skalierung (linear zwischen 5 und 50 für bessere Sichtbarkeit) ---
s_min, s_max = 5, 50
df_idx['Bubble_Size'] = np.interp(df_idx['Land_area'], (df_idx['Land_area'].min(), df_idx['Land_area'].max()), (s_min, s_max))

# --- 5. Visualisierung ---
fig = go.Figure()

# Scatter Plot (Bubbles)
fig.add_trace(go.Scatter(
    x=df_idx['Combined_Potential_Index'],
    y=df_idx['School_density'],
    mode='markers+text',
    text=df_idx['Country'],
    textposition="top center",
    marker=dict(
        size=df_idx['Bubble_Size'],
        color=df_idx['Combined_Potential_Index'],
        colorscale='Viridis',
        showscale=True,
        colorbar=dict(title="Index Score"),
        line=dict(width=1, color='white')
    ),
    hovertemplate="<b>%{text}</b><br>Index: %{x:.3f}<br>Dichte: %{y:.2f}<extra></extra>"
))

# Finale Regression für die Master-Linie
slope, intercept, r_master, p_master, _ = stats.linregress(df_idx['Combined_Potential_Index'], df_idx['School_density'])
line_x = np.array([0, 1])
line_y = slope * line_x + intercept

fig.add_trace(go.Scatter(
    x=line_x, y=line_y,
    mode='lines',
    name=f'Trend (R={r_master:.3f})',
    line=dict(color='red', dash='dash', width=2)
))

fig.update_layout(
    title='<b>Master Correlation: Combined Potential vs. Actual School Density</b><br>' +
          f'<sup>Weights derived from current dataset. Total R = {r_master:.3f}</sup>',
    xaxis_title='Combined Potential Index (Wealth, Equality, Human Capital, Compactness)',
    yaxis_title='School Density (per 100k)',
    template='plotly_white',
    height=850,
    width=1100
)

fig.show()

# Die Top-Ausreißer nach oben (Overperformer)
df_idx['Efficiency_Score'] = df_idx['School_density'] - (slope * df_idx['Combined_Potential_Index'] + intercept)
print("\nTop 5 Länder, die ihr Potenzial am besten nutzen (Overperformer):")
print(df_idx.sort_values(by='Efficiency_Score', ascending=False)[['Country', 'Efficiency_Score']].head(5))
```

```python
import plotly.graph_objects as go
import numpy as np
from scipy import stats
from sklearn.preprocessing import MinMaxScaler

# --- 1. Vorbereitung der Daten-Gruppen ---
# Gruppe A: Deine validen Länder
df_valid = df_meaningful.copy()

# Gruppe B: Alle anderen (Blockierer & Länder mit fehlenden Werten)
# Wir finden sie, indem wir alle aus df_analysis nehmen, die nicht in df_valid sind
df_excluded = df_analysis[~df_analysis['Country'].isin(df_valid['Country'])].copy()

# --- 2. Index-Logik (Berechnung basierend auf df_valid) ---
# Gewichte berechnen (wie zuvor)
factors = ['GDP_per_capita', 'Gini_index', 'Human_capital_index_plus', 'Land_area']
weights = {}
for col in factors:
    x_val = np.log10(df_valid[col]) if col in ['GDP_per_capita', 'Land_area'] else df_valid[col]
    r, _ = stats.pearsonr(x_val, df_valid['School_density'])
    weights[col] = abs(r)

# Skalierung initialisieren (fit nur auf validen Daten!)
scaler = MinMaxScaler()
def prepare_features(df):
    temp = df.copy()
    # Fehlende Werte in den ausgeschlossenen Ländern füllen wir mit dem Median der validen Länder 
    # (damit sie im Graph platziert werden können)
    for col in factors:
        temp[col] = temp[col].fillna(df_valid[col].median())
    
    temp['Gini_inv'] = df_valid['Gini_index'].max() - temp['Gini_index']
    temp['Area_inv'] = np.log10(df_valid['Land_area']).max() - np.log10(temp['Land_area'])
    temp['GDP_log'] = np.log10(temp['GDP_per_capita'].replace(0, 1)) # Vermeide log(0)
    return temp

df_valid_proc = prepare_features(df_valid)
df_excl_proc = prepare_features(df_excluded)

# Skalieren
cols_to_scale = ['GDP_log', 'Gini_inv', 'Human_capital_index_plus', 'Area_inv']
df_valid_proc[cols_to_scale] = scaler.fit_transform(df_valid_proc[cols_to_scale])
df_excl_proc[cols_to_scale] = scaler.transform(df_excl_proc[cols_to_scale]) # Transform nutzt die Skala von 'valid'

# Index berechnen
total_w = sum(weights.values())
for df in [df_valid_proc, df_excl_proc]:
    df['Combined_Potential_Index'] = (
        (df['GDP_log'] * weights['GDP_per_capita'] +
         df['Gini_inv'] * weights['Gini_index'] +
         df['Human_capital_index_plus'] * weights['Human_capital_index_plus'] +
         df['Area_inv'] * weights['Land_area']) / total_w
    )

# Bubble Size (5 bis 50)
all_areas = df_analysis['Land_area']
def get_size(series):
    return np.interp(series, (all_areas.min(), all_areas.max()), (5, 50))

df_valid_proc['Bubble_Size'] = get_size(df_valid_proc['Land_area'])
df_excl_proc['Bubble_Size'] = get_size(df_excl_proc['Land_area'])

# --- 3. Visualisierung ---
fig = go.Figure()

# A. Graue Bubbles (Ausgeschlossene Länder)
fig.add_trace(go.Scatter(
    x=df_excl_proc['Combined_Potential_Index'],
    y=df_excl_proc['School_density'],
    mode='markers',
    name='Systemic Outliers / Incomplete Data',
    text=df_excl_proc['Country'],
    marker=dict(
        size=df_excl_proc['Bubble_Size'],
        color='rgba(200, 200, 200, 0.4)', # Dezentes Grau
        line=dict(width=0.5, color='gray')
    ),
    hovertemplate="<b>%{text} (Excluded)</b><br>Index: %{x:.3f}<br>Dichte: %{y:.2f}<extra></extra>"
))

# B. Farbige Bubbles (Meaningful Data)
fig.add_trace(go.Scatter(
    x=df_valid_proc['Combined_Potential_Index'],
    y=df_valid_proc['School_density'],
    mode='markers+text',
    name='Meaningful Data',
    text=df_valid_proc['Country'],
    textposition="top center",
    marker=dict(
        size=df_valid_proc['Bubble_Size'],
        color=df_valid_proc['Combined_Potential_Index'],
        colorscale='Viridis',
        showscale=True,
        colorbar=dict(title="Index Score"),
        line=dict(width=1.2, color='white')
    ),
    hovertemplate="<b>%{text}</b><br>Index: %{x:.3f}<br>Dichte: %{y:.2f}<extra></extra>"
))

# C. Trendlinie (nur basierend auf validen Daten)
slope, intercept, r_master, _, _ = stats.linregress(df_valid_proc['Combined_Potential_Index'], df_valid_proc['School_density'])
line_x = np.array([0, 1])
line_y = slope * line_x + intercept

fig.add_trace(go.Scatter(
    x=line_x, y=line_y,
    mode='lines',
    name=f'Valid Trend (R={r_master:.3f})',
    line=dict(color='firebrick', dash='dash', width=3)
))

# Layout
fig.update_layout(
    title='<b>The Global School Infrastructure Gap</b><br>' +
          f'<sup>Valid Trendline excluding structural blockers. Grey bubbles indicate systemic outliers. Total N={len(df_analysis)}</sup>',
    xaxis_title='Socio-Economic Potential Index (0 = Low, 1 = Optimal)',
    yaxis_title='Actual School Density (per 100k)',
    template='plotly_white',
    height=900, width=1300,
    legend=dict(yanchor="top", y=0.99, xanchor="left", x=0.01)
)

fig.show()
```

```python
import wbgapi as wb
import pandas as pd
import plotly.express as px
import numpy as np

# 1. Mapping Definition
region_mapping = {
    'SSF': 'Sub-Saharan Africa', 'Z7': 'Sub-Saharan Africa',
    'EAS': 'East Asia & Pacific', 'Z4': 'East Asia & Pacific',
    'MEA': 'Middle East & North Africa', 'Z1': 'Middle East & North Africa',
    'ECS': 'Europe & Central Asia', 'T7': 'Europe & Central Asia',
    'SAS': 'South Asia', 'S3': 'South Asia',
    'NAC': 'North America', 'X4': 'North America',
    'LCN': 'Latin America & Caribbean', 'ZJ': 'Latin America & Caribbean'
}

try:
    # 2. Weltbank Metadaten laden
    economy_info = wb.economy.DataFrame().reset_index()
    economy_info = economy_info[['id', 'region']].rename(columns={'id': 'ISO3', 'region': 'Continent_Code'})

    # 3. Aufräumen: Alle potenziell doppelten Spalten aus final_df löschen
    cols_to_remove = ['Continent', 'Continent_Code', 'Continent_Code_x', 'Continent_Code_y']
    for col in cols_to_remove:
        if col in final_df.columns:
            final_df = final_df.drop(columns=[col])

    # 4. Zusammenführen
    final_df = pd.merge(final_df, economy_info, on='ISO3', how='left')

    # 5. Mapping anwenden
    final_df['Continent'] = final_df['Continent_Code'].map(region_mapping).fillna(final_df['Continent_Code'])
    final_df.loc[final_df['Country_Clean'] == 'Taiwan', 'Continent'] = 'East Asia & Pacific'

    # Filter für den Graphen
    mask = (final_df['GDP_per_Capita_2022'].notna() & (final_df['Count'] > 10))
    plot_df = final_df[mask].copy()

    # 6. Bubble Chart mit Trendlinie
    # Hinweis: statsmodels muss installiert sein für das Konfidenzintervall!
    fig = px.scatter(
        plot_df,
        x="GDP_per_Capita_2022",
        y="Count_Per_Mil",
        size="Count",
        color="Continent",
        hover_name="Country_Clean",
        log_x=True,
        size_max=60,
        trendline="ols", 
        trendline_scope="overall",
        trendline_color_override="black",
        # trendline_options mit add_constant ist wichtig für die statistische Korrektheit
        trendline_options=dict(add_constant=True),
        title="Montessori-Dichte vs. Wirtschaftskraft (mit 95% Konfidenzintervall)",
        labels={"GDP_per_Capita_2022": "BIP pro Kopf (USD)", "Count_Per_Mil": "Schulen pro Mio."},
        template="plotly_white"
    )

    # 7. Den Schatten (Konfidenzintervall) erzwingen
    # Wir iterieren durch die Traces, um die unsichtbaren Intervalle zu stylen
    for trace in fig.data:
        # Plotly Express nennt die CI-Traces oft nach der Trendlinie oder lässt den Namen leer
        if hasattr(trace, 'fill') and (trace.fill == 'tonexty' or trace.fill == 'toself'):
            trace.fillcolor = 'rgba(180, 180, 180, 0.4)' # Sichtbares Grau
            trace.line.color = 'rgba(180, 180, 180, 0)'   # Keine harten Ränder
            trace.name = "95% Konfidenzintervall"

    fig.update_layout(
        height=800,
        legend=dict(itemclick=False, itemdoubleclick=False)
    )
    fig.show()

except Exception as e:
    print(f"Fehler: {e}")
```

```python
import wbgapi as wb
import pandas as pd
import plotly.graph_objects as go
import numpy as np
from sklearn.linear_model import LinearRegression

# 1. Datenvorbereitung (Mapping & Merge wie gehabt)
region_mapping = {'SSF': 'Sub-Saharan Africa', 'Z7': 'Sub-Saharan Africa', 'EAS': 'East Asia & Pacific', 'Z4': 'East Asia & Pacific', 'MEA': 'Middle East & North Africa', 'Z1': 'Middle East & North Africa', 'ECS': 'Europe & Central Asia', 'T7': 'Europe & Central Asia', 'SAS': 'South Asia', 'S3': 'South Asia', 'NAC': 'North America', 'X4': 'North America', 'LCN': 'Latin America & Caribbean', 'ZJ': 'Latin America & Caribbean'}

try:
    economy_info = wb.economy.DataFrame().reset_index()[['id', 'region']].rename(columns={'id': 'ISO3', 'region': 'Continent_Code'})
    for col in ['Continent', 'Continent_Code', 'Continent_Code_x']:
        if col in final_df.columns: final_df.drop(columns=[col], inplace=True)
    
    final_df = pd.merge(final_df, economy_info, on='ISO3', how='left')
    final_df['Continent'] = final_df['Continent_Code'].map(region_mapping).fillna(final_df['Continent_Code'])
    
    plot_df = final_df[final_df['GDP_per_Capita_2022'].notna() & (final_df['Count'] > 10)].copy()
    plot_df = plot_df.sort_values('GDP_per_Capita_2022') # Wichtig für die Schatten-Zeichnung

    # --- MANUELLE BERECHNUNG DES TRENDS & KONFIDENZINTERVALLS ---
    x = np.log10(plot_df['GDP_per_Capita_2022'].values).reshape(-1, 1)
    y = plot_df['Count_Per_Mil'].values
    
    model = LinearRegression().fit(x, y)
    y_pred = model.predict(x)
    
    # Einfache Schätzung der Standardabweichung (für den visuellen Schatten)
    sum_err = np.sum((y - y_pred)**2)
    stdev = np.sqrt(sum_err / (len(y) - 2))
    # Wir nehmen 1.96 für das 95% Konfidenzintervall
    y_upper = y_pred + (1.96 * stdev / np.sqrt(len(y))) * 5 # Faktor 5 zur besseren Sichtbarkeit der Varianz
    y_lower = y_pred - (1.96 * stdev / np.sqrt(len(y))) * 5

    # --- PLOT MIT GRAPH OBJECTS (für volle Kontrolle) ---
    fig = go.Figure()

    # 1. Der Schatten (Konfidenzintervall)
    fig.add_trace(go.Scatter(
        x=np.concatenate([plot_df['GDP_per_Capita_2022'], plot_df['GDP_per_Capita_2022'][::-1]]),
        y=np.concatenate([y_upper, y_lower[::-1]]),
        fill='toself',
        fillcolor='rgba(200, 200, 200, 0.3)',
        line=dict(color='rgba(255,255,255,0)'),
        hoverinfo="skip",
        name="Konfidenzbereich (geschätzt)"
    ))

    # 2. Die Trendlinie
    fig.add_trace(go.Scatter(
        x=plot_df['GDP_per_Capita_2022'], y=y_pred,
        line=dict(color='black', width=2),
        name="Trendlinie"
    ))

    # 3. Die Bubbles
    for continent in plot_df['Continent'].unique():
        cont_df = plot_df[plot_df['Continent'] == continent]
        fig.add_trace(go.Scatter(
            x=cont_df['GDP_per_Capita_2022'], y=cont_df['Count_Per_Mil'],
            mode='markers',
            marker=dict(size=cont_df['Count'], sizemode='area', sizeref=2.*max(plot_df['Count'])/(60.**2)),
            name=continent,
            text=cont_df['Country_Clean'],
            hovertemplate="<b>%{text}</b><br>BIP: %{x}<br>Dichte: %{y}<extra></extra>"
        ))

    fig.update_xaxes(type="log", title="BIP pro Kopf (USD)")
    fig.update_yaxes(title="Schulen pro Mio. Einwohner")
    fig.update_layout(title="Montessori Trend & Varianz", height=800, template="plotly_white")
    fig.show()

except Exception as e:
    print(f"Fehler: {e}")
```

```python
def get_latest_wb_data_with_year(indicator_code, col_name):
    """
    Holt den MRV UND das zugehörige Jahr für einen Indikator.
    """
    print(f"Abfrage läuft: {indicator_code}...")
    results = []
    
    try:
        # fetch liefert uns Zugriff auf das Feld 'time'
        for row in wb.data.fetch(indicator_code, mrnev=1):
            results.append({
                'ISO3': row['economy'],
                col_name: row['value'],
                f"{col_name}_Year": row['time']
            })
        
        return pd.DataFrame(results)
    
    except Exception as e:
        print(f"Fehler bei {indicator_code}: {e}")
        return pd.DataFrame(columns=['ISO3', col_name, f"{col_name}_Year"])

# --- ANWENDUNG ---

# 1. Gini-Daten laden
df_gini_smart = get_latest_wb_data_with_year('SI.POV.GINI', 'Gini_Index')
```

```python
# Falls noch nicht installiert: !pip install pytrends
from pytrends.request import TrendReq
import matplotlib.pyplot as plt

# 1. Verbindung zu Google aufbauen
# hl='de-DE' (Host Language), tz=360 (Timezone Offset)
pytrends = TrendReq(hl='de-DE', tz=360)

# 2. Parameter definieren
kw_list = ["Montessori"]
# Zeitformat: 'YYYY-MM-DD YYYY-MM-DD'
timeframe = '2021-01-01 2025-12-31'
geo = 'CN' # Hier kannst du jedes ISO2-Kürzel einsetzen

# 3. API-Abfrage senden
print(f"Lade Google Trends für {kw_list} in {geo}...")
pytrends.build_payload(kw_list, cat=0, timeframe=timeframe, geo=geo, gprop='')

# 4. Daten abrufen (Interest Over Time)
df_trends = pytrends.interest_over_time()

# 5. Visualisierung
if not df_trends.empty:
    plt.figure(figsize=(12, 6))
    plt.plot(df_trends.index, df_trends['Montessori'], label='Suchinteresse', color='blue')
    plt.title(f'Google Trends: Interesse an "Montessori" in {geo} (2021-2025)')
    plt.xlabel('Datum')
    plt.ylabel('Relatives Interesse (0-100)')
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.show()
else:
    print(f"Keine Daten für {geo} im gewählten Zeitraum gefunden.")
```

```python
from sklearn.preprocessing import MinMaxScaler

# 1. Daten normalisieren (0 bis 1), damit wir sie verrechnen können
scaler = MinMaxScaler()
cols_to_index = ['Human_Capital_Index', 'Trend_Score', 'GDP_per_Capita_2022', 'Gini_Index']
scaled_values = scaler.fit_transform(final_df[cols_to_index])
scaled_df = pd.DataFrame(scaled_values, columns=cols_to_index, index=final_df.index)

# 2. Gewichtung basierend auf deinen Korrelationen (Betrag nehmen)
# Wir nutzen deine Werte: HCI (0.49), Gini (0.28), Trend (0.16)
# Für das BIP nehmen wir einen geschätzten Wert (z.B. 0.35), falls nicht berechnet
weights = {
    'Human_Capital_Index': 0.49,
    'Gini_Index': 0.28,      # Wir "drehen" den Gini später um
    'Trend_Score': 0.16,
    'GDP_per_Capita_2022': 0.35 
}

# Gewichte normalisieren, damit sie in Summe 1 ergeben
total_w = sum(weights.values())
w = {k: v/total_w for k, v in weights.items()}

# 3. Den Index berechnen
final_df['Readiness_Index'] = (
    scaled_df['Human_Capital_Index'] * w['Human_Capital_Index'] +
    scaled_df['Trend_Score'] * w['Trend_Score'] +
    scaled_df['GDP_per_Capita_2022'] * w['GDP_per_Capita_2022'] +
    (1 - scaled_df['Gini_Index']) * w['Gini_Index'] # 1-Gini, weil niedriger Gini besser ist
)
```

```python
from sklearn.preprocessing import MinMaxScaler
import statsmodels.api as sm

# 1. Normalisierung der Komponenten
scaler = MinMaxScaler()
cols_to_index = ['Human_Capital_Index', 'Trend_Score', 'GDP_per_Capita_2022', 'Gini_Index']

# Wir erstellen eine Kopie für die Skalierung, um den originalen final_df nicht zu verändern
scaled_values = scaler.fit_transform(final_df[cols_to_index].fillna(final_df[cols_to_index].median()))
scaled_df = pd.DataFrame(scaled_values, columns=cols_to_index, index=final_df.index)

# 2. Gewichteter Index (MRI)
# Gewichte: HCI (0.49), Gini (0.28), BIP (0.35), Trend (0.16) -> normalisiert auf Summe 1
weights = {'Human_Capital_Index': 0.38, 'Gini_Index': 0.22, 'GDP_per_Capita_2022': 0.27, 'Trend_Score': 0.13}

final_df['Readiness_Index'] = (
    scaled_df['Human_Capital_Index'] * weights['Human_Capital_Index'] +
    scaled_df['Trend_Score'] * weights['Trend_Score'] +
    scaled_df['GDP_per_Capita_2022'] * weights['GDP_per_Capita_2022'] +
    (1 - scaled_df['Gini_Index']) * weights['Gini_Index']
)

# 3. Gap-Berechnung (Differenz zwischen Potential und Realität)
# Wir berechnen, wie weit ein Land unter der Ideal-Linie liegt
final_df['Opportunity_Gap'] = final_df['Readiness_Index'] - (final_df['Count_Per_Mil'] / final_df['Count_Per_Mil'].max())
```

```python
import matplotlib.pyplot as plt
import numpy as np
import statsmodels.api as sm
import adjustText

# 1. Data preparation for the trendline
# Sorting by Readiness_Index ensures the line and interval are drawn correctly
df_sorted = plot_data.sort_values('Readiness_Index')
X = sm.add_constant(df_sorted['Readiness_Index'])
y = df_sorted['Count_Per_Mil']

# Calculate Regression
model = sm.OLS(y, X).fit()
predictions = model.get_prediction(X)
summary_frame = predictions.summary_frame(alpha=0.05) # 95% Confidence Interval
r_squared = model.rsquared

# 2. Create Plot
plt.figure(figsize=(14, 10))

# Draw Confidence Interval (the "shadow")
plt.fill_between(
    df_sorted['Readiness_Index'], 
    summary_frame['mean_ci_lower'], 
    summary_frame['mean_ci_upper'], 
    color='gray', alpha=0.15, 
    label='95% Confidence Interval' # CHANGED to English
)

# Trendline (the red dashed line)
plt.plot(
    df_sorted['Readiness_Index'], 
    summary_frame['mean'], 
    color='red', linestyle='--', linewidth=2, 
    label=f'Regression Trend (R²={r_squared:.2f})' # CHANGED to English + added R2
)

# Scatter Plot
# Note: Point size (s) is based on Population_2022
scatter = plt.scatter(
    plot_data['Readiness_Index'], 
    plot_data['Count_Per_Mil'], 
    s=plot_data['Population_2022']/1_000_000 * 15, # Adjusted scaling for Matplotlib
    c=plot_data['Opportunity_Gap'], 
    cmap='RdYlGn', 
    alpha=0.3, 
    edgecolors="w",
    zorder=3
)

# Add Country Names
texts = []
for i, row in plot_data.iterrows():
    texts.append(plt.text(row['Readiness_Index'], row['Count_Per_Mil'], row['Country_Clean_x'], fontsize=10))

# Arrange names to avoid overlap
adjustText.adjust_text(texts, arrowprops=dict(arrowstyle='->', color='black', lw=0.5))

# Details & Styling (All labels changed to English)
# Main Title (Large)
plt.title('Market Potential Analysis: Readiness vs. School Density\nCircle size represents total population (2022)', fontsize=11, pad=20, color='#555555')


plt.xlabel('Montessori Readiness Index (Potential)', fontsize=12)
plt.ylabel('Schools per 1 Million Inhabitants', fontsize=12)

# Colorbar for the Opportunity Gap
cbar = plt.colorbar(scatter)
cbar.set_label('Opportunity Gap (Market Potential)', fontsize=12)

plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='upper left')

# Optimize Layout
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
df_waldorf = pd.concat([tables[0].df, tables[1].df], ignore_index=True)
```

```python
df_waldorf = df_waldorf.rename(columns={
    0: "Country", 
    1: "Kindergarden_Count", 
    2: "School_Count"
})
```

```python
# 2. Zeilen löschen, in denen "Total" im Country vorkommt
# Wir behalten nur die Zeilen, bei denen die Bedingung NICHT zutrifft (~)
df_waldorf = df_waldorf[~df_waldorf['Country'].str.contains('Total', case=False, na=False)]

# 3. Zeilen löschen, in denen "Kinder" im Kindergarden_Count vorkommt
# na=False verhindert Fehler, falls leere Zellen vorhanden sind
df_waldorf = df_waldorf[~df_waldorf['Kindergarden_Count'].astype(str).str.contains('Kinder', case=False, na=False)]
# 1. Ländernamen bereinigen: Falls '/', nimm den zweiten Teil, sonst den ersten
# .str.split('/').str[-1] nimmt das letzte Element (funktioniert bei 'A/B' -> 'B' und bei 'A' -> 'A')
df_waldorf['Country'] = df_waldorf['Country'].str.split('/').str[-1].str.strip()
# 1. Zuerst alle Blanks (Leerzeichen) entfernen
df_waldorf['Kindergarden_Count'] = df_waldorf['Kindergarden_Count'].astype(str).str.strip()

# 2. Zeilen behalten, die weder NaN noch ein leerer String sind
# Wir filtern: Nicht leerer String UND Nicht Null
df_waldorf = df_waldorf[
    (df_waldorf['Kindergarden_Count'] != "") & 
    (df_waldorf['Kindergarden_Count'].notna()) &
    (df_waldorf['Kindergarden_Count'] != "nan") # Falls durch astype(str) 'nan' entstanden is
    ]

# 3. Index zur Sauberkeit zurücksetzen
df_waldorf = df_waldorf.reset_index(drop=True)
# Optional: Index zurücksetzen, da Zeilen gelöscht wurden
df_waldorf = df_waldorf.reset_index(drop=True)
```

```python
# 1. Den "*" entfernen (in beiden Spalten)
# regex=False stellt sicher, dass der Stern als normales Zeichen behandelt wird
df_waldorf['Kindergarden_Count'] = df_waldorf['Kindergarden_Count'].astype(str).str.replace('*', '', regex=False)
df_waldorf['School_Count'] = df_waldorf['School_Count'].astype(str).str.replace('*', '', regex=False)

# 2. Leere Strings (die durch das Entfernen des * entstehen könnten) durch 0 ersetzen oder behandeln
# (Optional, falls ein Feld nur aus "*" bestand)
df_waldorf['Kindergarden_Count'] = df_waldorf['Kindergarden_Count'].replace('', '0')
df_waldorf['School_Count'] = df_waldorf['School_Count'].replace('', '0')

# 3. In Integers umwandeln
# Wir nutzen pd.to_numeric mit downcast='integer' für eine saubere Konvertierung
df_waldorf['Kindergarden_Count'] = pd.to_numeric(df_waldorf['Kindergarden_Count']).astype(int)
df_waldorf['School_Count'] = pd.to_numeric(df_waldorf['School_Count']).astype(int)
```

```python
import country_converter as coco

# Initialisiere den Converter
cc = coco.CountryConverter()

# ISO3-Codes basierend auf der Spalte 'Country' erstellen
df_waldorf['ISO3'] = cc.convert(names=df_waldorf['Country'], to='ISO3')
```
